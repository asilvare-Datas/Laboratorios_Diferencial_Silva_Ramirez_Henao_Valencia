## 5 · La pregunta importante: ¿el consumo es una *función* del mes?

Vamos a graficar **la nube completa**: un punto por cada pareja (mes, consumo de un hogar). Antes de correr la celda, relean su respuesta a P3.

```python
# La nube completa (si hay muchos puntos, muestreamos 20.000 para que la gráfica respire).
muestra = df.sample(min(20000, len(df)), random_state=1)

meses = sorted(df["mes"].unique())
pos = {m: i for i, m in enumerate(meses)}   # posición de cada mes en el eje x

plt.figure(figsize=(10, 4.5))
plt.scatter(muestra["mes"].map(pos), muestra["kwh"], s=3, alpha=0.15)
plt.xticks(range(0, len(meses), 3), meses[::3], rotation=45)
plt.xlabel("Mes")
plt.ylabel("Consumo del hogar (kWh)")
plt.title("Todos los hogares, todos los meses: la ciudad entera en una nube")

# La recta vertical del test, dibujada sobre un mes cualquiera:
plt.axvline(pos[meses[len(meses)//2]], color="red", linewidth=1.5, linestyle="--",
            label="Recta vertical del test")
plt.legend()
plt.show()

un_mes = meses[len(meses)//2]
print(f"¿Cuántos consumos distintos tiene el mes {un_mes}?:",
      df[df['mes'] == un_mes]['kwh'].nunique())
```

🗣️ **La recta roja atraviesa la nube ¿en cuántos puntos?** Entonces: sobre toda la ciudad, ¿el consumo es función del mes? Escriban aquí su veredicto **usando la definición formal de la clase** (dominio, codominio, regla):

El Veredicto: El consumo NO es función del mes.

Dominio: El conjunto de meses del estudio {2011-12, ..., 2014-02}
Codominio: Los consumos posibles [0, M] ⊂ ℝ (kWh)
Regla: A cada mes le asociamos el consumo de un hogar.

Para que fuera función, cada mes del dominio debería tener exactamente un consumo asociado pero en 2013-01 hay 5.496 registros, con 5.258 consumos distintos entre 0 y 3.349,66 kWh. La recta vertical roja atraviesa miles de puntos de la nube, no uno. Entonces la regla no determina un único valor. Dado el mes, no sabemos cuál de los miles de consumos corresponde, falta una variable que lo decida.


Ahora viene el movimiento clave. Vamos a **condicionar**: en lugar de toda la ciudad, un solo hogar.

```python
# ✏️ Ejercicio 4 — Un solo hogar.
# Elijan un código de hogar (mírenlos con df["hogar"].unique()[:10]) y grafiquen SU serie:
# mes en el eje x, kwh en el eje y. ¿Cuántos puntos atraviesa ahora una recta vertical?

hogar_elegido = df["hogar"].unique()[0]   # ← cambien por el hogar que quieran
serie = df[df["hogar"] == hogar_elegido].sort_values("mes")

plt.figure(figsize=(10, 4))
plt.plot(serie["mes"].map(pos), serie["kwh"], marker="o", markersize=4)
plt.xticks(range(0, len(meses), 3), meses[::3], rotation=45)
plt.xlabel("Mes")
plt.ylabel("Consumo (kWh)")
plt.title(f"La historia eléctrica del hogar {hogar_elegido}")
plt.show()

print("Registros de este hogar:", len(serie),
      "| ¿Algún mes repetido?:", serie["mes"].duplicated().sum())
```

🗣️ **Conclusión del movimiento (la frase más importante del laboratorio):**

Al condicionar por hogar, la relación **se volvió** función, porque: *(explíquenlo con dominio, codominio y regla)*

- **Dominio:** ya no son los meses de toda la ciudad, sino los meses en que el hogar elegido tiene lectura, D_h ⊆ {2011-12, …, 2014-02}.
- **Codominio:** el mismo de antes, [0, M] ⊂ ℝ (kWh).
- **Regla:** a cada mes de D_h le asignamos el consumo que registró *ese* hogar en *ese* mes.

Antes, dado un mes, había que elegir entre miles de consumos (en 2013-01, 5.258 valores distintos) porque la variable `hogar` estaba escondida y variaba libremente. Al fijarla, la regla sí decide: la recta vertical corta la serie del hogar en un solo punto, y el conteo de meses repetidos de ese hogar dio 0.

Frase obligatoria del informe — complétenla con sus palabras: **“Una relación deja de ser función cuando…”**

## 6 · Momento 3 — Explicar

🗣️ Media página del equipo, aquí mismo: ¿qué predijeron (P1, P2, P3)? ¿Qué encontraron? ¿Cuál fue la discrepancia más interesante y a qué la atribuyen?

**Qué predijimos.** En P1 estimamos 350 kWh/mes para un hogar londinense, por el frío y el uso constante de calefacción. Creíamos que consumiría más que uno manizaleño. En P2 esperábamos una curva en forma de U o V, con los meses más altos en diciembre–febrero. En P3 respondimos que el consumo no es función del mes, porque depende del tamaño de la casa, los hábitos y el número de personas.

**Qué encontramos.** La mediana real fue de unos 245 kWh/mes (la mitad de los registros está por debajo). El histograma es asimétrico, con una cola larga que llega a más de 3.000 kWh. La curva mediana por mes sí tiene forma de U: picos en diciembre y enero, y valles en verano. La nube completa confirmó P3: la recta vertical atraviesa miles de puntos en un mismo mes, así que el consumo no es función del mes. Al filtrar un solo hogar, cada mes tiene un único consumo.

**Discrepancias.** La más grande fue P1: sobrestimamos la mediana en más de 100 kWh. Lo atribuimos a que supusimos que toda la calefacción de Londres es eléctrica; probablemente una parte importante usa gas, que este dataset no mide (es una hipótesis, no la probamos). Además, pensar en un "hogar típico" con un número nos hizo ignorar la dispersión: hay hogares muy por encima de la mediana. La segunda fue febrero: esperábamos que estuviera alto, pero cae por debajo de enero y marzo. Sospechamos que se debe a que febrero tiene menos días, lo que vuelve poco justa la unidad "kWh por mes" (la pista de la tarea puente).

**Volvimos a P3.** Mantenemos el "no", pero ahora con definición formal: el consumo es función de (hogar, mes), no solo del mes.

**Una relación deja de ser función cuando** a un mismo elemento del dominio le corresponden varios valores, porque hay una variable que no estamos teniendo en cuenta.
